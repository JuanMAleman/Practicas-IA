```python
#Pregunta 2.4
import json, subprocess, sys
import numpy as np

res = {"MLP": [], "CNN": []}
for s in [1, 2, 3, 4, 5]:
    out = f"outputs/seed_{s}"
    subprocess.run([sys.executable, "cnn_lab_scaffold.py", "--part", "baselines",
                    "--epochs", "3", "--seed", str(s), "--out", out, "--subset", "1000"],
                   check=True, capture_output=True)
    r = json.load(open(f"{out}/02_baselines.json"))
    for k in res:
        res[k].append(r[k]["val_acc"])

for k, v in res.items():
    print(f"{k}: media={np.mean(v):.4f}  std={np.std(v, ddof=1):.4f}  valores={np.round(v, 4)}")
```
