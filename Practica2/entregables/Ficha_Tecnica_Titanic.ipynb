# Ficha Técnica: Dataset Titanic

### 1. Identificación del dataset.
Nombre: Titanic: Aprendizaje automático a partir del desastre.

Fuente: Keaggle

Licencias y términos de uso: https://www.kaggle.com/competitions/titanic/overview

Archivo: Train.csv

Unidad de análisis: Cada fila, un pasajero del Titanic.

Dimensiones: 891 filas y 12 columnas.

### 2. Descripción de variables.
Variable Objetivo: "Survived" (0 = no sobrevivió, 1 = sobrevivió).

Numéricas: Age, SibSp, Parch, Fare.

Categóricas: Survived, Pclass, Sex, Embarked, Cabin.

Identificadores/texto: PassengerId, Name, Ticket.

### 3. Diagnóstico de calidad de los datos.
- Cabin ausente en 77.10 % y con 147 valores distintos.
- Age ausente en 19.87 %. Embarked ausente en 0.22 %.
- Fare con cola larga (media 32.20, mediana 14.45, máximo 512.33) y 15 valores en 0.
- Desbalance leve de la clase objetivo (61.62 % vs. 38.38 %; IR ≈ 1.61).
- Sin filas duplicadas, sin valores fuera de dominio, sin categorías inconsistentes.

### 4. Limitaciones
- Datos históricos de una muestra de pasajeros; no generalizan a otros eventos.
- Los identificadores (PassengerId, Name, Ticket) no aportan información predictiva y se excluyen del modelo.
